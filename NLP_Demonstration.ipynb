# Focus AI Pro: NLP Core Algorithm Demonstration

This notebook demonstrates the core **Natural Language Processing (NLP)** algorithm utilized inside the Focus AI Pro extension. 

To maximize privacy, completely eliminate latency, and avoid relying on expensive external APIs (like Gemini or OpenAI), our production code uses a completely localized **Lexicon-Based Bag-of-Words** approach executed natively in JavaScript.

This Jupyter Notebook runs the **Python equivalent** of that exact algorithm to demonstrate to judges exactly how our Machine Learning polarity matrix evaluates distraction vs. productivity.

```python
import re

# 1. Define our Sentiment Lexicon Matrices (Knowledge Base)
# In the extension, these evaluate the emotional "Vibe" and "Toxicity" of the video titles heavily consumed by the user.
positive_lexicon = r'(positive|focus|study|learn|code|build|success|motivat|happy|love|relax|chill|tutorial|guide|course|develop)'
negative_lexicon = r'(drama|angry|sad|fail|hate|worst|scary|horror|war|news|depress|fight|doom|anxiety|stress|bad|toxic)'

# 2. Sample Data: Simulated clickstream of a user's recent video titles
watch_history = [
    "How to build a machine learning model in 10 minutes",
    "Worst drama of the century! So toxic...",
    "Study with me: 4 hours of deep focus",
    "10 scary horror stories that will give you anxiety",
    "Complete guide to React development"
]

print("Starting NLP Clickstream Evaluation...\n")

total_score = 0

# 3. Execute NLP Matching Algorithm
for title in watch_history:
    title_score = 0
    if re.search(positive_lexicon, title, re.IGNORECASE):
        title_score += 1
        total_score += 1
    if re.search(negative_lexicon, title, re.IGNORECASE):
        title_score -= 1
        total_score -= 1
        
    # Output classification per title
    classification = "Positive/Educational" if title_score > 0 else "Negative/Stressful" if title_score < 0 else "Neutral"
    print(f"Video: '{title}'")
    print(f" -> Classification: {classification} (Score: {title_score})\n")
```

### Conclusion Rule Engine
After aggregating the emotional polarity vector array for the entire session, the engine classifies the overarching psychological vibe of the user's focus state.

```python
print(f"\nFinal Aggregated Session Vibe Score: {total_score}")

if total_score > 1:
    print("Diagnosis: High-performance consumption. Displaying 'Positive Trend' insights to user.")
elif total_score < -1:
    print("Diagnosis: Doomscrolling detected. Displaying 'Stress Warning' to user.")
else:
    print("Diagnosis: Emotionally neutral consumption pattern.")
```
